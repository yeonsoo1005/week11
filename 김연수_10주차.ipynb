```python
# 1. 클래스 정의 셀 다시 실행
class prep_class:
    def __init__(self, df, id_col):
        self.df = df
        self.id_col = id_col

        print('=' * 20, '데이터프레임 중 일부 데이터를 출력합니다.', '=' * 20)
        print(self.df.shape)
        display(self.df.head())

        print('\n', '=' * 20, '데이터프레임의 info를 출력합니다.', '=' * 20)
        print(self.df.info())

        print('\n', '=' * 20, '데이터프레임의 describe를 출력합니다.', '=' * 20)
        display(self.df.describe())

    def transform_melt(self, var_name, value_name, value_col_list):
        print('=' * 20, '변환 이전의 데이터프레임 형태입니다.', '=' * 20)
        print(self.df.shape)
        display(self.df.head())

        melted_df = self.df.melt(
            id_vars=self.id_col,
            value_vars=value_col_list,
            var_name=var_name,
            value_name=value_name
        )

        melted_df = melted_df.sort_values(by=self.id_col)

        print('\n' + '=' * 20, '변환 이후의 데이터프레임 형태입니다.', '=' * 20)
        print(melted_df.shape)
        display(melted_df.head())

        self.df = melted_df
```

```python
import pandas as pd
employees_df = pd.read_csv('sampledatabase/employees.csv')
customeres_df = pd.read_csv('sampledatabase/customers.csv')

print('='*10, 'employees.csv 정보입니다.', '='*10)
display(employees_df.shape)
display(employees_df.head())

print('='*10, 'customers.csv 정보입니다.', '='*10)
display(customeres_df.shape)
display(customeres_df.head())
```

<img src="./참고이미지/Q4/Q4_1.png" width="70%">

<b>코드 Cell</b>
- employees 데이터에 대한 전처리 과정입니다.
- 직속 상사의 LastName, 직속 상사의 직급, 직속 부하의 LastName, 직속 부하의 직급, 직속부하의 직원번호 데이터를 정리합니다.
- 데이터를 적절히 분할하고 결합하는 과정이 필요할 수 있습니다.
- 셀 출력에서는 데이터를 간편하게 보기 위해, 컬럼 이름 변경, 컬럼 순서 변경 등의 작업을 수행했습니다.

```python
print('=' * 20, '컬럼을 정리하는 중간 과정입니다.', '=' * 20)

reporter_df = employees_df[['lastName', 'jobTitle', 'employeeNumber', 'reportsTo']].copy()
reporter_df.columns = ['Reporter_LastName', 'Reporter_JobTitle', 'Reporter_EmpNumber', 'reportsTo']

boss_df = employees_df[['employeeNumber', 'lastName', 'jobTitle']].copy()
boss_df.columns = ['Boss_EmpNumber', 'BossLastName', 'Boss_JobTitle']

emp_df = pd.merge(
    reporter_df,
    boss_df,
    left_on='reportsTo',
    right_on='Boss_EmpNumber',
    how='left'
)

display(emp_df.shape)
display(emp_df.head())
print(emp_df.columns)

print('=' * 20, '직속 상사, 직속 부하에 대한 정리 결과입니다.', '=' * 20)

result_df = emp_df[
    ['BossLastName', 'Boss_JobTitle', 'Reporter_LastName', 'Reporter_JobTitle', 'Reporter_EmpNumber']
].dropna()

display(result_df.shape)
display(result_df.head())
```

<img src="./참고이미지/Q4/Q4_2.png" width="45%">

<b>코드 Cell</b>
- 위에서 정리한 데이터 정보와 customer.csv 정보를 결합합니다.
- salesRepEmployeeNumber에 적혀있는 직원번호가 해당 고객을 관리하는 직원의 직원번호입니다.

```python
print('=' * 20, '전처리 데이터와 customers_df 데이터를 결합합니다.', '=' * 20)

merged_df = pd.merge(
    result_df,
    customeres_df,
    left_on='Reporter_EmpNumber',
    right_on='salesRepEmployeeNumber',
    how='left'
)

print(merged_df.columns)
display(merged_df.shape)
display(merged_df.head(10))

print('=' * 20, '필요한 컬럼 정보만 남겨 데이터를 정리합니다.', '=' * 20)

merged_df = merged_df[
    ['BossLastName', 'Boss_JobTitle',
     'Reporter_LastName', 'Reporter_JobTitle', 'Reporter_EmpNumber',
     'salesRepEmployeeNumber', 'customerName', 'creditLimit']
]

display(merged_df.shape)
display(merged_df.head(20))
```

<img src="./참고이미지/Q4/Q4_3.png" width="100%">

<b>코드 Cell</b>
- 최종 결과입니다.
- 직속 부하가 관리하는 고객들의 creditLimit의 평균값을 계산합니다.
- 집계한 creditLimit 정보를 기준으로 내림차순 정렬을 수행한 뒤에, 순위를 매깁니다.

```python
final_df = (
    merged_df.dropna(subset=['creditLimit'])
    .groupby(['BossLastName', 'Boss_JobTitle'])['creditLimit']
    .mean()
    .reset_index()
)

final_df.columns = ['BossLastName', 'BOSS_jobTitle', 'creditLimit']
final_df = final_df.sort_values(by='creditLimit', ascending=False)
final_df['RANK'] = final_df['creditLimit'].rank(method='dense', ascending=False).astype(int)
final_df = final_df[['RANK', 'BossLastName', 'BOSS_jobTitle', 'creditLimit']]
final_df = final_df.reset_index(drop=True)

display(final_df)
```

```python
print('=' * 10, 'customers 데이터입니다.', '=' * 10)

customers_df = customers_df[['customerNumber', 'country']]

display(customers_df.shape)
display(customers_df.head())

customers_df = customers_df[customers_df['country'] == 'USA']

print('=' * 10, 'country에서 USA만 남긴 상태입니다.', '=' * 10)

display(customers_df.shape)
display(customers_df.head())
```

<b>Cell 설명</b>
- 전처리 데이터들을 결합합니다.
- 최종적으로, 주문상태는 'Shipped', customer의 country는 USA, 주문별 total_order_value 정보를 포함한 데이터가 만들어집니다.

```python

merged_df = pd.merge(
    orders_df,
    total_order_df,
    on='orderNumber',
    how='left'
)


merged_df = pd.merge(
    merged_df,
    customers_df,
    on='customerNumber',
    how='inner'
)

display(merged_df.shape)
display(merged_df.head())
```

<img src="./참고이미지/Q6/Q6_6.png" width="45%">

<b>최종 결과</b>
- customerNumber 별로 총주문금액의 최대, 최소, 최대값과 최소값의 차이를 나타냅니다.
- 해당 결과를 최대값과 최소값의 차이를 기준으로 내림차순으로 정렬한 이후에, 순위를 부여합니다.

```python
final_df = (
    merged_df
    .groupby('customerNumber')['total_order_value']
    .agg(['max', 'min'])
    .reset_index()
)

final_df.columns = [
    'customerNumber',
    'order_value_max',
    'order_value_min'
]

final_df['order_value_diff'] = (
    final_df['order_value_max'] - final_df['order_value_min']
)

final_df = final_df.sort_values(
    by='order_value_diff',
    ascending=False
)

final_df = final_df.head(20)

final_df['RANK'] = final_df['order_value_diff'].rank(
    method='dense',
    ascending=False
).astype(int)

final_df = final_df[
    ['RANK', 'customerNumber',
     'order_value_max', 'order_value_min', 'order_value_diff']
].reset_index(drop=True)


display(final_df.shape)
display(final_df)
```

# Q7. 데이터 전처리 - Python MySQL

<b>참고 사항</b>
- 반드시, 하나의 쿼리문을 통해 동일한 결과를 도출해야 합니다.
- 쿼리를 두 번에 나누어서 적용할 경우, 오답으로 처리될 수 있습니다.

```python

import mysql.connector

conn = mysql.connector.connect(
    host='localhost',      
    user='root',           
    password='yeonDB1005',    
    database='classicmodels',  
    port=3306              
)
```

<b>쿼리문 작성 Cell</b>
- 설명
    - 쿼리문을 작성하여 결과를 출력해주세요.

- 참고사항
    - 해당 셀에서는 반드시 하나의 쿼리문만 작성되어야 하며 데이터 출력 결과를 위해 별도의 쿼리 혹은 코드가 추가된 경우 오답으로 처리될 수 있습니다.

```python
query = """
SELECT 
    CAST(@rank := @rank + 1 AS UNSIGNED) AS `RANK`,
    t.customerNumber,
    t.order_value_max,
    t.order_value_min,
    t.order_value_diff
FROM (
    SELECT 
        o.customerNumber,
        MAX(od_sum.total_order_value) AS order_value_max,
        MIN(od_sum.total_order_value) AS order_value_min,
        MAX(od_sum.total_order_value) - MIN(od_sum.total_order_value) AS order_value_diff
    FROM orders o
    JOIN (
        SELECT 
            orderNumber,
            SUM(quantityOrdered * priceEach) AS total_order_value
        FROM orderdetails
        GROUP BY orderNumber
    ) od_sum
        ON o.orderNumber = od_sum.orderNumber
    JOIN customers c
        ON o.customerNumber = c.customerNumber
    WHERE 
        o.status = 'Shipped'
        AND c.country = 'USA'
    GROUP BY o.customerNumber
    ORDER BY order_value_diff DESC
    LIMIT 20
) t,
(SELECT @rank := 0) r
;
"""

pd.read_sql(query, conn)
```
